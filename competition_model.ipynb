```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from catboost import CatBoostClassifier
import numpy as np

TARGET = "default payment next month"

X = train.drop(columns=[TARGET, "ID"]).copy()
y = train[TARGET].copy()

# These are categorical/discrete variables
cat_cols = [
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

cat_indices = [X.columns.get_loc(col) for col in cat_cols]

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cat_scores = []

for fold, (train_idx, valid_idx) in enumerate(cv.split(X, y), 1):

    X_train = X.iloc[train_idx]
    X_valid = X.iloc[valid_idx]

    y_train = y.iloc[train_idx]
    y_valid = y.iloc[valid_idx]

    model = CatBoostClassifier(
        iterations=1000,
        learning_rate=0.03,
        depth=6,
        loss_function="Logloss",
        eval_metric="AUC",
        l2_leaf_reg=5,
        random_seed=42,
        verbose=False,
        allow_writing_files=False,
        thread_count=-1
    )

    model.fit(
        X_train,
        y_train,
        cat_features=cat_indices,
        eval_set=(X_valid, y_valid),
        early_stopping_rounds=100,
        verbose=False
    )

    valid_pred = model.predict_proba(X_valid)[:, 1]

    score = roc_auc_score(y_valid, valid_pred)
    cat_scores.append(score)

    print(
        f"Fold {fold}: "
        f"ROC-AUC = {score:.6f} | "
        f"Best iteration = {model.get_best_iteration()}"
    )

print("\n" + "=" * 50)
print(f"CatBoost Mean ROC-AUC: {np.mean(cat_scores):.6f}")
print(f"CatBoost Std:          {np.std(cat_scores):.6f}")
print("=" * 50)
```

```python
import numpy as np
import pandas as pd

TARGET = "default payment next month"


def make_features(df):
    df = df.copy()

    # ---------------------------------------------------------
    # 1. Original repayment-status columns
    # ---------------------------------------------------------
    pay_cols = [
        "PAY_0", "PAY_2", "PAY_3",
        "PAY_4", "PAY_5", "PAY_6"
    ]

    bill_cols = [
        "BILL_AMT1", "BILL_AMT2", "BILL_AMT3",
        "BILL_AMT4", "BILL_AMT5", "BILL_AMT6"
    ]

    pay_amt_cols = [
        "PAY_AMT1", "PAY_AMT2", "PAY_AMT3",
        "PAY_AMT4", "PAY_AMT5", "PAY_AMT6"
    ]

    # ---------------------------------------------------------
    # 2. Repayment behavior aggregates
    # ---------------------------------------------------------
    df["PAY_MEAN"] = df[pay_cols].mean(axis=1)
    df["PAY_STD"] = df[pay_cols].std(axis=1)
    df["PAY_MIN"] = df[pay_cols].min(axis=1)
    df["PAY_MAX"] = df[pay_cols].max(axis=1)

    # Number of months with delayed payment
    df["PAY_DELAY_COUNT"] = (df[pay_cols] > 0).sum(axis=1)

    # Serious delay
    df["PAY_SEVERE_COUNT"] = (df[pay_cols] >= 2).sum(axis=1)

    # Any serious delay
    df["PAY_SEVERE_ANY"] = (df[pay_cols] >= 2).max(axis=1).astype(int)

    # Maximum historical delay
    df["PAY_MAX_DELAY"] = df[pay_cols].max(axis=1)

    # Recent repayment behavior
    df["PAY_RECENT_MEAN"] = df[
        ["PAY_0", "PAY_2", "PAY_3"]
    ].mean(axis=1)

    df["PAY_OLDER_MEAN"] = df[
        ["PAY_4", "PAY_5", "PAY_6"]
    ].mean(axis=1)

    df["PAY_RECENT_VS_OLD"] = (
        df["PAY_RECENT_MEAN"] -
        df["PAY_OLDER_MEAN"]
    )

    # ---------------------------------------------------------
    # 3. Repayment trend
    # ---------------------------------------------------------
    df["PAY_TREND"] = (
        df["PAY_0"] - df["PAY_6"]
    )

    df["PAY_RECENT_WORSE"] = (
        df["PAY_0"] > df["PAY_6"]
    ).astype(int)

    # Consecutive/current delay indicators
    df["CURRENT_DELAY"] = (df["PAY_0"] > 0).astype(int)

    df["CURRENT_SEVERE_DELAY"] = (
        df["PAY_0"] >= 2
    ).astype(int)

    # ---------------------------------------------------------
    # 4. Bill amount statistics
    # ---------------------------------------------------------
    df["BILL_MEAN"] = df[bill_cols].mean(axis=1)
    df["BILL_STD"] = df[bill_cols].std(axis=1)
    df["BILL_MIN"] = df[bill_cols].min(axis=1)
    df["BILL_MAX"] = df[bill_cols].max(axis=1)

    df["BILL_RANGE"] = (
        df["BILL_MAX"] - df["BILL_MIN"]
    )

    # Recent bill behavior
    df["BILL_RECENT_MEAN"] = df[
        ["BILL_AMT1", "BILL_AMT2", "BILL_AMT3"]
    ].mean(axis=1)

    df["BILL_OLD_MEAN"] = df[
        ["BILL_AMT4", "BILL_AMT5", "BILL_AMT6"]
    ].mean(axis=1)

    df["BILL_RECENT_CHANGE"] = (
        df["BILL_RECENT_MEAN"] -
        df["BILL_OLD_MEAN"]
    )

    # ---------------------------------------------------------
    # 5. Bill changes month-to-month
    # ---------------------------------------------------------
    for i in range(1, 6):
        current = bill_cols[i - 1]
        previous = bill_cols[i]

        df[f"BILL_DIFF_{i}"] = (
            df[current] - df[previous]
        )

    # ---------------------------------------------------------
    # 6. Payment amount statistics
    # ---------------------------------------------------------
    df["PAY_AMT_MEAN"] = df[pay_amt_cols].mean(axis=1)
    df["PAY_AMT_STD"] = df[pay_amt_cols].std(axis=1)
    df["PAY_AMT_MIN"] = df[pay_amt_cols].min(axis=1)
    df["PAY_AMT_MAX"] = df[pay_amt_cols].max(axis=1)

    df["PAY_AMT_TOTAL"] = df[pay_amt_cols].sum(axis=1)

    df["PAY_AMT_NONZERO_COUNT"] = (
        df[pay_amt_cols] > 0
    ).sum(axis=1)

    # ---------------------------------------------------------
    # 7. Credit utilization
    # ---------------------------------------------------------
    limit = df["LIMIT_BAL"].replace(0, np.nan)

    for col in bill_cols:
        df[f"{col}_UTIL"] = (
            df[col] / limit
        )

    df["AVG_UTILIZATION"] = (
        df[bill_cols].mean(axis=1) / limit
    )

    df["MAX_UTILIZATION"] = (
        df[bill_cols].max(axis=1) / limit
    )

    # ---------------------------------------------------------
    # 8. Payment relative to bills
    # ---------------------------------------------------------
    for bill_col, payment_col in zip(
        bill_cols,
        pay_amt_cols
    ):
        bill = df[bill_col].abs() + 1
        df[f"{payment_col}_TO_BILL"] = (
            df[payment_col] / bill
        )

    # ---------------------------------------------------------
    # 9. Payment coverage
    # ---------------------------------------------------------
    df["TOTAL_PAYMENT_TO_TOTAL_BILL"] = (
        df["PAY_AMT_TOTAL"] /
        (df[bill_cols].abs().sum(axis=1) + 1)
    )

    # ---------------------------------------------------------
    # 10. Age / credit-limit interactions
    # ---------------------------------------------------------
    df["LIMIT_PER_AGE"] = (
        df["LIMIT_BAL"] /
        (df["AGE"] + 1)
    )

    # ---------------------------------------------------------
    # 11. Clean infinite values
    # ---------------------------------------------------------
    df.replace(
        [np.inf, -np.inf],
        np.nan,
        inplace=True
    )

    df.fillna(0, inplace=True)

    return df


# Create engineered datasets
X_fe = make_features(
    train.drop(columns=[TARGET])
)

y = train[TARGET].copy()

print("Original features :", train.shape[1] - 1)
print("New feature count :", X_fe.shape[1])

print("\nNew features:")
print(
    [
        c for c in X_fe.columns
        if c not in train.drop(columns=[TARGET]).columns
    ]
)
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from catboost import CatBoostClassifier
import numpy as np

TARGET = "default payment next month"

# Remove ID because it is an identifier, not a customer feature
X_fe_model = X_fe.drop(columns=["ID"], errors="ignore")
y = train[TARGET].copy()

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

fe_scores = []
best_iterations = []

for fold, (train_idx, valid_idx) in enumerate(cv.split(X_fe_model, y), 1):

    X_train = X_fe_model.iloc[train_idx]
    X_valid = X_fe_model.iloc[valid_idx]

    y_train = y.iloc[train_idx]
    y_valid = y.iloc[valid_idx]

    model = CatBoostClassifier(
        iterations=1500,
        learning_rate=0.03,
        depth=6,
        loss_function="Logloss",
        eval_metric="AUC",
        l2_leaf_reg=5,
        random_seed=42,
        verbose=False,
        allow_writing_files=False,
        thread_count=-1
    )

    model.fit(
        X_train,
        y_train,
        eval_set=(X_valid, y_valid),
        early_stopping_rounds=150,
        verbose=False
    )

    valid_pred = model.predict_proba(X_valid)[:, 1]

    score = roc_auc_score(
        y_valid,
        valid_pred
    )

    fe_scores.append(score)

    best_iter = model.get_best_iteration()
    best_iterations.append(best_iter)

    print(
        f"Fold {fold}: "
        f"ROC-AUC = {score:.6f} | "
        f"Best iteration = {best_iter}"
    )

print("\n" + "=" * 60)
print("CATBOOST + FEATURE ENGINEERING")
print("=" * 60)

print(
    f"Mean ROC-AUC: {np.mean(fe_scores):.6f}"
)

print(
    f"Std:          {np.std(fe_scores):.6f}"
)

print(
    f"Best iterations: {best_iterations}"
)

print("=" * 60)
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import roc_auc_score
from catboost import CatBoostClassifier
from lightgbm import LGBMClassifier
import lightgbm as lgb
import numpy as np
import pandas as pd

TARGET = "default payment next month"

X_blend = X_fe.drop(columns=["ID"], errors="ignore").copy()
y_blend = train[TARGET].copy()

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# ---------------------------------------------------------
# OOF prediction arrays
# ---------------------------------------------------------

oof_cat = np.zeros(len(train))
oof_lgb = np.zeros(len(train))

# ---------------------------------------------------------
# CatBoost categorical columns
# ---------------------------------------------------------

cat_cols = [
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

cat_indices = [
    X_blend.columns.get_loc(col)
    for col in cat_cols
]

# ---------------------------------------------------------
# Cross-validation
# ---------------------------------------------------------

for fold, (train_idx, valid_idx) in enumerate(
    cv.split(X_blend, y_blend), 1
):

    X_train = X_blend.iloc[train_idx]
    X_valid = X_blend.iloc[valid_idx]

    y_train = y_blend.iloc[train_idx]
    y_valid = y_blend.iloc[valid_idx]

    # =====================================================
    # CATBOOST
    # =====================================================

    cat_model = CatBoostClassifier(
        iterations=1500,
        learning_rate=0.03,
        depth=6,
        loss_function="Logloss",
        eval_metric="AUC",
        l2_leaf_reg=5,
        random_seed=42,
        verbose=False,
        allow_writing_files=False,
        thread_count=-1
    )

    cat_model.fit(
        X_train,
        y_train,
        cat_features=cat_indices,
        eval_set=(X_valid, y_valid),
        early_stopping_rounds=150,
        verbose=False
    )

    cat_pred = cat_model.predict_proba(X_valid)[:, 1]

    oof_cat[valid_idx] = cat_pred

    cat_score = roc_auc_score(
        y_valid,
        cat_pred
    )

    # =====================================================
    # LIGHTGBM
    # =====================================================

    lgb_model = LGBMClassifier(
        n_estimators=1500,
        learning_rate=0.03,
        num_leaves=31,
        max_depth=-1,
        min_child_samples=30,
        subsample=0.8,
        colsample_bytree=0.8,
        reg_alpha=0.1,
        reg_lambda=2.0,
        objective="binary",
        random_state=42,
        n_jobs=-1,
        verbosity=-1
    )

    lgb_model.fit(
        X_train,
        y_train,
        eval_set=[(X_valid, y_valid)],
        callbacks=[
            lgb.early_stopping(
                150,
                verbose=False
            )
        ]
    )

    lgb_pred = lgb_model.predict_proba(X_valid)[:, 1]

    oof_lgb[valid_idx] = lgb_pred

    lgb_score = roc_auc_score(
        y_valid,
        lgb_pred
    )

    print(
        f"Fold {fold}: "
        f"CatBoost = {cat_score:.6f} | "
        f"LightGBM = {lgb_score:.6f}"
    )


# ---------------------------------------------------------
# Individual OOF scores
# ---------------------------------------------------------

cat_oof_score = roc_auc_score(
    y_blend,
    oof_cat
)

lgb_oof_score = roc_auc_score(
    y_blend,
    oof_lgb
)

print("\n" + "=" * 65)
print("OOF MODEL PERFORMANCE")
print("=" * 65)

print(f"CatBoost OOF ROC-AUC : {cat_oof_score:.6f}")
print(f"LightGBM OOF ROC-AUC : {lgb_oof_score:.6f}")
```

```python
import numpy as np
import pandas as pd

TARGET = "default payment next month"

def create_features(df):
    df = df.copy()

    # -----------------------------
    # Payment status features
    # -----------------------------
    pay_cols = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]

    df["PAY_MEAN"] = df[pay_cols].mean(axis=1)
    df["PAY_STD"] = df[pay_cols].std(axis=1)
    df["PAY_MIN"] = df[pay_cols].min(axis=1)
    df["PAY_MAX"] = df[pay_cols].max(axis=1)

    df["PAY_DELAY_COUNT"] = (df[pay_cols] > 0).sum(axis=1)
    df["PAY_SEVERE_COUNT"] = (df[pay_cols] >= 2).sum(axis=1)
    df["PAY_SEVERE_ANY"] = (df[pay_cols] >= 2).any(axis=1).astype(int)
    df["PAY_MAX_DELAY"] = df[pay_cols].max(axis=1)

    # Recent vs older payment behavior
    recent_pay = ["PAY_0", "PAY_2", "PAY_3"]
    older_pay = ["PAY_4", "PAY_5", "PAY_6"]

    df["PAY_RECENT_MEAN"] = df[recent_pay].mean(axis=1)
    df["PAY_OLDER_MEAN"] = df[older_pay].mean(axis=1)

    df["PAY_RECENT_VS_OLD"] = (
        df["PAY_RECENT_MEAN"] - df["PAY_OLDER_MEAN"]
    )

    df["PAY_TREND"] = (
        df["PAY_0"] - df["PAY_6"]
    )

    df["PAY_RECENT_WORSE"] = (
        df["PAY_RECENT_MEAN"] > df["PAY_OLDER_MEAN"]
    ).astype(int)

    df["CURRENT_DELAY"] = (df["PAY_0"] > 0).astype(int)
    df["CURRENT_SEVERE_DELAY"] = (df["PAY_0"] >= 2).astype(int)

    # -----------------------------
    # Bill amount features
    # -----------------------------
    bill_cols = [
        "BILL_AMT1",
        "BILL_AMT2",
        "BILL_AMT3",
        "BILL_AMT4",
        "BILL_AMT5",
        "BILL_AMT6"
    ]

    df["BILL_MEAN"] = df[bill_cols].mean(axis=1)
    df["BILL_STD"] = df[bill_cols].std(axis=1)
    df["BILL_MIN"] = df[bill_cols].min(axis=1)
    df["BILL_MAX"] = df[bill_cols].max(axis=1)
    df["BILL_RANGE"] = df["BILL_MAX"] - df["BILL_MIN"]

    recent_bill = ["BILL_AMT1", "BILL_AMT2", "BILL_AMT3"]
    old_bill = ["BILL_AMT4", "BILL_AMT5", "BILL_AMT6"]

    df["BILL_RECENT_MEAN"] = df[recent_bill].mean(axis=1)
    df["BILL_OLD_MEAN"] = df[old_bill].mean(axis=1)

    df["BILL_RECENT_CHANGE"] = (
        df["BILL_RECENT_MEAN"] - df["BILL_OLD_MEAN"]
    )

    # Month-to-month bill changes
    for i in range(1, 6):
        df[f"BILL_DIFF_{i}"] = (
            df[f"BILL_AMT{i}"] -
            df[f"BILL_AMT{i+1}"]
        )

    # -----------------------------
    # Payment amount features
    # -----------------------------
    pay_amt_cols = [
        "PAY_AMT1",
        "PAY_AMT2",
        "PAY_AMT3",
        "PAY_AMT4",
        "PAY_AMT5",
        "PAY_AMT6"
    ]

    df["PAY_AMT_MEAN"] = df[pay_amt_cols].mean(axis=1)
    df["PAY_AMT_STD"] = df[pay_amt_cols].std(axis=1)
    df["PAY_AMT_MIN"] = df[pay_amt_cols].min(axis=1)
    df["PAY_AMT_MAX"] = df[pay_amt_cols].max(axis=1)
    df["PAY_AMT_TOTAL"] = df[pay_amt_cols].sum(axis=1)
    df["PAY_AMT_NONZERO_COUNT"] = (
        df[pay_amt_cols] > 0
    ).sum(axis=1)

    # -----------------------------
    # Credit utilization
    # -----------------------------
    for i in range(1, 7):
        df[f"BILL_AMT{i}_UTIL"] = (
            df[f"BILL_AMT{i}"] /
            (df["LIMIT_BAL"] + 1)
        )

    util_cols = [
        f"BILL_AMT{i}_UTIL"
        for i in range(1, 7)
    ]

    df["AVG_UTILIZATION"] = df[util_cols].mean(axis=1)
    df["MAX_UTILIZATION"] = df[util_cols].max(axis=1)

    # -----------------------------
    # Payment-to-bill ratios
    # -----------------------------
    for i in range(1, 7):
        df[f"PAY_AMT{i}_TO_BILL"] = (
            df[f"PAY_AMT{i}"] /
            (df[f"BILL_AMT{i}"].abs() + 1)
        )

    df["TOTAL_PAYMENT_TO_TOTAL_BILL"] = (
        df["PAY_AMT_TOTAL"] /
        (df[bill_cols].abs().sum(axis=1) + 1)
    )

    # -----------------------------
    # Credit limit relative to age
    # -----------------------------
    df["LIMIT_PER_AGE"] = (
        df["LIMIT_BAL"] /
        (df["AGE"] + 1)
    )

    # Convert booleans to integers
    bool_cols = df.select_dtypes(include=["bool"]).columns
    df[bool_cols] = df[bool_cols].astype(int)

    return df


# Create engineered dataset
X_fe = create_features(
    train.drop(columns=[TARGET])
)

y = train[TARGET].copy()

print("=" * 60)
print("FEATURE ENGINEERING COMPLETE")
print("=" * 60)
print("Original features :", train.shape[1] - 1)
print("New feature count  :", X_fe.shape[1])
print("Missing values     :", X_fe.isnull().sum().sum())
print(
    "Infinite values    :",
    np.isinf(
        X_fe.select_dtypes(include=np.number)
    ).sum().sum()
)

print("\nShape:", X_fe.shape)
```
